 ## My lane as an ML task

**Task type: Classification**

My chosen lane is Refresh / Content Opportunity Scoring. ## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

## Target or proxy

The starter dataset does not contain a separate `is_declining_label` column. It contains `trend_direction`, which represents the observed current trend of each content item.

For this framing exercise, I use `trend_direction` as a proxy for identifying content that may need refresh attention. This is an observed current-trend proxy, not a future refresh-success outcome. A stronger target for a later ML model would be an outcome measured in a future time window.

Because `trend_direction` and `trend_pct` contain the trend information used for this proxy, they should not be used as predictive features when modeling the target.

```python
# Check the observed trend categories
print("Observed trend direction counts:")
print(df["trend_direction"].value_counts(dropna=False))
```

## Success metric

The main success metric is **Precision@50**. It measures the proportion of relevant content items among the top 50 items prioritized for review. A higher Precision@50 means that the review queue contains more relevant content opportunities, helping the team focus limited editorial effort on higher-priority pages.

```python
# Define the review queue size
K = 50

print("K for Precision@K:", K)
print("Total content items:", len(df))
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

## Why ML beats a fixed rule here

A fixed rule might prioritize content using only one condition, such as low CTR or a certain level of impressions. However, content opportunity can depend on multiple signals, including impressions, CTR, average position, engagement, content age, freshness, and content characteristics. These signals may interact in ways that are difficult to capture with a single if-statement.

Machine learning can combine multiple signals to identify patterns and produce a prioritized review queue. The output would support human decision-making rather than automatically deciding which content must be refreshed. The ML approach should be compared with a simple rule-based baseline before claiming that it performs better.
